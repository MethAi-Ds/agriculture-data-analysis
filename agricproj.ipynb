```python
# EXPLORATORY THRESHOLD
#LOW VS HIGH TEMPERATURE: CROP YIELD ANALYSIS USING THRESHOLD 
GROUPA = methy[(methy['avg_temp'] >= 1) & (methy['avg_temp'] <=17)]
GROUPB = methy[(methy['avg_temp'] >= 17.1) & (methy['avg_temp'] <=31)]
THRESH_H = methy['hg/ha_yield'].quantile(0.98)
print(THRESH_H)
print('GROUPA =', GROUPA.value_counts().count(),'is the total number of avg_temp that is less than 17')
print('GROUPB =', GROUPB.value_counts().count(),'is the total number of avg_temp that is greater than 17 to the highest number of avg_temp in the dataset')
```

```python
k = GROUPA['hg/ha_yield']>= THRESH_H
print('GROUPA RESULT:',k.value_counts())
print('')
r = GROUPB['hg/ha_yield']>= THRESH_H
print('GROUPB RESULT:',r.value_counts())
```

## EXPLORATORY THRESHOLD EXPLANATION
TO FURTHER INVESTIGATE THE EFFECT OF TEMPERATURE ON YIELD,
I DECIDED TO SET A YIELD THRESHOLD OF 98TH PERCENTILE(319534.84hg/ha) WHCIH REPRESENT THE TOP 2% HIGHEST YIELDS IN THE DATASET

THEN I SPLIT THE AVG_TEMP INTO TWO GROUPS, GROUPA  AND GROUPB.
  GROUPA CONTAINS AVG_TEMP FROM (1°C-17°C )WHICH HAS TOTAL NUMBER OF 4143 RECORDS AND
  GROUPB CONTAINS AVG_TEMP FROM (17.1°C TO THE HIGHEST AVG_TEMP 31°C) WHICH HAS THE TOTAL NUMBER OF 8918 RECORDS

THEN I COMPARED EACH GROUP AGAINST THE THRESHOLD.
THE RESULT WAS

GROUPA = 216(5.2%) EXCEEDED OUT OF 4143 RECORDS

GROUPB = ONLY 43(0.48%) EXCEEDED OUT OF 8918 RECORDS

DESPITE GROUPB HAVING MORE RECORDS, GROUPA PRODUCES SIGNIFICANTLY MORE HIGH YIELD CASES.
   THESE SUGGESTS THAT LOWER AVG_TEMP ARE ASSOCIATED WITH EXTREMELY HIGH CROP YIELDS.
